# EDA and visualization
Discover class counts, image sizes, and representative samples.

```python
from pathlib import Path
import matplotlib.pyplot as plt
from src.data_loader import collect_paths
paths, labels = collect_paths('../casting_512x512')
print({name: int((labels == i).sum()) for i, name in enumerate(('OK', 'Defective'))})
fig, axes = plt.subplots(1, 6, figsize=(15, 3))
for axis, path, label in zip(axes, paths[:6], labels[:6]):
    axis.imshow(plt.imread(path), cmap='gray'); axis.set_title('Defective' if label else 'OK'); axis.axis('off')
plt.show()
```
